frequent = []
with open("frequent_customers.txt", "r") as file:
    for line in file:
        frequent.append(line.strip())

for name in todays_visitors:

        print(f"{name} is a frequent customer! ⭐")

### Part 2

```python
visit_counts = {}
for name in frequent:
    visit_counts[name] = 0

with open("visits_last_week.txt", "r") as file:
    for line in file:
        day, customer = line.strip().split(",")
        if customer in visit_counts:
            visit_counts[customer] += 1

with open("visit_report.txt", "w") as report:
    report.write("Frequent Customer Visits - Last Week\n")
    report.write("-" * 36 + "\n")
    for name, count in visit_counts.items():
        report.write(f"{name}: {count}\n")
    report.write("-" * 36 + "\n")
    report.write(f"Total visits: {sum(visit_counts.values())}\n")

with open("visit_report.txt", "r") as report:
    print(report.read())
```
